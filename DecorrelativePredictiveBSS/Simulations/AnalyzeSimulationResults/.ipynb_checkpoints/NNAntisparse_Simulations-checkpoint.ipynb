```python
def create_summary_table(df):
    """
    Creates a summary table with Mean ± Std for SINR and Execution Time.
    """
    # Group by Model and calculate mean and std
    summary = df.groupby('Model').agg({
        'SINR': ['mean', 'std'],
        'execution_time': ['mean', 'std']
    })

    # Flatten the MultiIndex columns
    summary.columns = ['SINR_mean', 'SINR_std', 'Time_mean', 'Time_std']

    # Create the formatted strings "Mean ± Std"
    final_table = pd.DataFrame(index=summary.index)
    
    final_table['SINR (dB)'] = summary.apply(
        lambda x: f"{x['SINR_mean']:.2f} ± {x['SINR_std']:.2f}", axis=1
    )
    
    final_table['Execution Time (s)'] = summary.apply(
        lambda x: f"{x['Time_mean']:.2f} ± {x['Time_std']:.2f}", axis=1
    )

    return final_table
```

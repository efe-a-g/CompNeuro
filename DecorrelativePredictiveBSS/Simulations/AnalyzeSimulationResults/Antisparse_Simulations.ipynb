```python
def create_summary_table(df):
    """
    Groups by Model and computes Mean ± Std.
    Explodes the SNR list to calculate true component-level statistics.
    """
    # 1. Calculate Mean Component SNR for each row
    # (Extracting from your list of lists format)
    df['mSNR'] = df['SNR'].apply(lambda x: np.mean(x[0]) if isinstance(x, list) else np.nan)

    # 2. Aggregate
    summary = df.groupby('Model').agg({
        'SINR': ['mean', 'std'],
        'mSNR': ['mean', 'std'],
        'execution_time': ['mean', 'std']
    })

    # 3. Format for Paper
    final = pd.DataFrame(index=summary.index)
    final['SINR (dB)'] = summary.apply(lambda x: f"{x[('SINR', 'mean')]:.2f} ± {x[('SINR', 'std')]:.2f}", axis=1)
    final['mSNR (dB)'] = summary.apply(lambda x: f"{x[('mSNR', 'mean')]:.2f} ± {x[('mSNR', 'std')]:.2f}", axis=1)
    final['Time (s)'] = summary.apply(lambda x: f"{x[('execution_time', 'mean')]:.2f} ± {x[('execution_time', 'std')]:.2f}", axis=1)

    return final
```
